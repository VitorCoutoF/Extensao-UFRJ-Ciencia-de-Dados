```python
import basedosdados as bd
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import display

# Preencha com o ID do seu projeto no Google Cloud (ex.: "seu-projeto-google-cloud")
BILLING_PROJECT_ID = "etlchamadosrj"

sns.set_theme(style="whitegrid")
print("Ambiente pronto! Execute a próxima célula para autenticar.")
```

```python
if not BILLING_PROJECT_ID:
    raise ValueError("Preencha o BILLING_PROJECT_ID na célula de configuração do notebook 00_autenticacao_bigquery.")

# Siga as instruções de login exibidas no primeiro acesso.
# SELECT 1 testa a conexão sem ler tabelas.
teste_autenticacao = bd.read_sql(
    "SELECT 1 AS conexao_ok",
    billing_project_id=BILLING_PROJECT_ID,
)
display(teste_autenticacao)
print(f"Autenticação concluída! Projeto: {BILLING_PROJECT_ID}")
```
